# Bài 7. Các hàm thống kê trên Tensor  
Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.  

```python
import torch
a = torch.randn(3,4)
print(a)
print("\nMean: ",torch.mean(a,dim=0))  # Tính trung bình theo từng cột
print("\nStandard Deviation: ",torch.std(a,dim=0))  # Tính độ lệch chuẩn theo từng cột
print("\nCumulative Sum: ", torch.cumsum(a, dim=0))  # Tổng tích lũy theo chiều dọc
```

# Yêu cầu:  
• Thay dim=0 bằng dim=1 và nhận xét sự khác nhau. 

```python
import torch
a = torch.randn(3,4)
print(a)
print("\nMean with dim=0: ",torch.mean(a,dim=0))  # Tính trung bình theo từng cột
print("\nMean with dim=1: ",torch.mean(a,dim=1))  # Tính trung bình theo từng hàng
print("\nStandard Deviation with dim=0: ",torch.std(a,dim=0))  # Tính độ lệch chuẩn theo từng cột
print("\nStandard Deviation with dim=1: ",torch.std(a,dim=1))  # Tính độ lệch chuẩn theo từng hàng
print("\nCumulative Sum with dim=0: ", torch.cumsum(a, dim=0))  # Tích lũy theo chiều dọc
print("\nCumulative Sum with dim=1: ", torch.cumsum(a, dim=1))  # Tích lũy theo chiều ngang
```

## Nhận xét

- Với `dim=0`, `mean()` và `std()` gộp các hàng để tính theo từng cột nên kết quả có shape `[4]`. `cumsum()` cộng dồn từ trên xuống dưới và vẫn giữ shape `[3, 4]`.
- Với `dim=1`, `mean()` và `std()` gộp các cột để tính theo từng hàng nên kết quả có shape `[3]`. `cumsum()` cộng dồn từ trái sang phải và vẫn giữ shape `[3, 4]`.

# Bài 8. Autograd cơ bản  
Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.

```python
import torch
x = torch.tensor(3.6, requires_grad=True)
y= x * x
y.backward()  # Tính đạo hàm tự động
print("x = ", x)
print("y = ", y)
print("x.grad = ", x.grad)  # In ra đạo hàm của y theo x
```

# Yêu cầu:  
• Tính thủ công đạo hàm của y = x² tại x = 3.6 và so sánh với x.grad
y = x²
y' = 2x 
Với x = 3.6, ta có y' = 2 * 3.6 = 7.2
So với x.grad = 7.2000 => Kết quả tính đạo hàm tự động bằng backward() là chính xác so với đạo hàm tính thủ công.
• Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient. 

```python
import torch

x = torch.tensor(3.6, requires_grad=True)
y = 3 * x ** 2 + 2 * x + 1
y.backward()
print("x = ", x)
print("y = ", y)
print("x.grad = ", x.grad)
```

# Bài 9. Gradient Descent tìm hệ số đa thức 
Sinh dữ liệu từ đa thức y = x² + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số. 

```python
import numpy as np 
import torch
polynomial = np.poly1d([1, 2, 3])  # Tạo đa thức y = x² + 2x + 3
N = 20
x = np.random.randn(N,1) * 5 
y = polynomial(x)
XX =np.hstack([x*x, x, np.ones_like(x)])
w = torch.randn(3,1,requires_grad=True)
x= torch.tensor(XX, dtype=torch.float32)
y= torch.tensor(y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial Coefficients: ", w)
for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w  
    mse = torch.mean(torch.square(y_pred - y))
    mse.backward()
    optimizer.step()
    if epoch % 100 == 0:
        print(f"Epoch {epoch}, MSE: {mse.item():.6f}, w: {w.detach().flatten()}")
```

# Yêu cầu:  
• So sánh ba hệ số học được với [1, 2, 3]
    Ba hệ số học được gần với [1, 2, 3] cụ thể với kết quả ở lần 900 với w :  w: tensor([1.0008, 2.0001, 2.9439]), điều này chứng tỏ rằng quá trình huấn luyện bằng NAdam và autograd đã thành công trong việc tìm lại các hệ số của đa thức gốc.
• Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step()
    optimizer.zero_grad(): Xóa các gradient đã tính toán trước đó để tránh việc cộng dồn gradient từ các bước huấn luyện trước.
    mse.backward(): Tính toán gradient của hàm mất mát (mean squared error) đối với các tham số của mô hình.
    optimizer.step(): Cập nhật các tham số của mô hình dựa trên gradient đã tính toán và thuật toán tối ưu hóa (NAdam trong trường hợp này).

# Bài 10. Giải hệ phương trình bằng Autograd  
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.
    Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1,   A+C=8, B-D=2

```python
import random
import torch
A = torch.tensor(random.random(),requires_grad=True)
B = torch.tensor(random.random(),requires_grad=True)
C = torch.tensor(random.random(),requires_grad=True)
D = torch.tensor(random.random(),requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)  
for epoch in range(EPOCHS):   
    y1 = A + B - 9   
    y2 = C - D - 1 
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()  
    sqerr.backward()  
    optimizer.step()  
    if epoch % 100 == 0:
        print(f"Epoch {epoch}, sqerr = {sqerr.item():.8f}")

# Kiểm tra nghiệm sau khi tối ưu
with torch.no_grad():
    errors = torch.stack([
        A + B - 9,
        C - D - 1,
        A + C - 8,
        B - D - 2
    ])

    print("\nNghiệm tìm được:")
    print(f"A = {A.item():.4f}")
    print(f"B = {B.item():.4f}")
    print(f"C = {C.item():.4f}")
    print(f"D = {D.item():.4f}")
    print("Các sai số:", errors)
    print("Tổng sai số bình phương:", torch.sum(errors**2).item())
```

#  Nhận xét vì sao bài toán có thể có nhiều nghiệm
Sau quá trình tối ưu, tổng sai số bình phương tiến gần về 0 và các giá trị A, B, C, D tìm được thỏa bốn phương trình.

Tuy nhiên, hệ không có nghiệm duy nhất vì phương trình thứ tư phụ thuộc vào ba phương trình còn lại. Nghiệm tổng quát của hệ là:

(A, B, C, D) = (t, 9-t, 8-t, 7-t)

Do đó, mỗi lần chạy chương trình có thể thu được một bộ giá trị khác nhau do khởi tạo ngẫu nhiên, nhưng chúng vẫn thỏa hệ nếu tổng sai số bình phương gần bằng 0.

### Kiểm tra đạo hàm bằng phép tính thủ công

Với $y=2x^2+5x+3$, ta có:

$$
\frac{dy}{dx}=4x+5.
$$

Tại $x=2$:

$$
\frac{dy}{dx}=4(2)+5=13.
$$

Kết quả `x.grad = 13` từ Autograd trùng với kết quả tính thủ công.

```python
# Thay đa thức trong Bài 9 bằng y = 2x² - 3x + 5; 
# huấn luyện và cho biết các hệ số tìm được. 
import numpy as np
import torch

np.random.seed(42)
torch.manual_seed(42)

polynomial = np.poly1d([2, -3, 5])  # y = 2x² - 3x + 5
N = 20
x_data = np.random.randn(N, 1) * 5
y_data = polynomial(x_data)
X_data = np.hstack([x_data**2, x_data, np.ones_like(x_data)])

X = torch.tensor(X_data, dtype=torch.float32)
y = torch.tensor(y_data, dtype=torch.float32)
w = torch.randn(3, 1, requires_grad=True)

optimizer = torch.optim.NAdam([w], lr=0.01)
EPOCHS = 5000

print("Hệ số ban đầu:", w.detach().flatten())
for epoch in range(EPOCHS):
    optimizer.zero_grad()
    y_pred = X @ w
    mse = torch.mean((y_pred - y) ** 2)
    mse.backward()
    optimizer.step()

    if epoch % 500 == 0:
        print(f"Epoch {epoch:4d}, MSE: {mse.item():.6f}, w: {w.detach().flatten()}")

learned = w.detach().flatten()
expected = torch.tensor([2.0, -3.0, 5.0])
print("\nHệ số tìm được:", learned)
print("Hệ số mong đợi:", expected)
print("Sai lệch tuyệt đối:", torch.abs(learned - expected))
```

## Tính thủ công các hệ số

Giả sử chưa biết các hệ số của đa thức và đặt:

$$
y=ax^2+bx+c.
$$

Chọn ba giá trị đơn giản $x=0$, $x=1$ và $x=2$ từ đa thức $y=2x^2-3x+5$:

- Với $x=0$: $y=5$, do đó $c=5$.
- Với $x=1$: $y=4$, do đó $a+b+c=4 \Rightarrow a+b=-1$.
- Với $x=2$: $y=7$, do đó $4a+2b+c=7 \Rightarrow 2a+b=1$.

Lấy $(2a+b)-(a+b)=1-(-1)$, ta được $a=2$. Thay vào $a+b=-1$:

$$
b=-1-2=-3.
$$

Vậy ba hệ số tính thủ công là:

$$
\boxed{[a,b,c]=[2,-3,5]}.
$$

Sau khi huấn luyện, các hệ số trong `w` phải xấp xỉ `[2, -3, 5]`. Chúng có thể sai lệch rất nhỏ vì NAdam tìm nghiệm bằng cách cập nhật lặp, thay vì giải hệ phương trình trực tiếp.
